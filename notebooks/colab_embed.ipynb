# Negative-Space Reader: encode an embedding bundle on a Colab GPU

Encodes the texts in `colab_bundle.zip` (made on the Mac by `embed.mode: export`, see `experiments/embed.py`) with the
same model and settings as the local store, and gives back `colab_out.zip` for `embed.mode: import`.

1. **Runtime → Change runtime type → T4 GPU** (the free tier is enough), then run the cells in order.
2. Cell 3 asks for `colab_bundle.zip`; cell 5 downloads `colab_out.zip`.

Nothing from the bundle is stored anywhere but this runtime; it is deleted when the runtime ends.
